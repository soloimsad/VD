```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Cargar el dataset
df = pd.read_csv("dataset/Survey_AI.csv")

# Configurar estilo
sns.set(style="whitegrid")
plt.figure(figsize=(10, 6))

# === Boxplot ===
sns.boxplot(data=df, x="Q13.Year_of_study", y="Q7.Utility_grade", palette="Set2")
plt.title("Boxplot: Confianza por Año de Estudio")
plt.xlabel("Año de Estudio")
plt.ylabel("Percepción de Utilidad de la IA")
plt.xticks(ticks=[0, 1], labels=["1°", "2°"])  # Solo si hay hasta 5 años

plt.tight_layout()
plt.show()
```
